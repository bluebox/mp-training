```python
from abc import ABC, abstractmethod
from datetime import datetime

class SmartDevice(ABC):
    _device_counter = 0

    def __init__(self, device_id):
        self._device_id = device_id
        self.__is_on = False
        SmartDevice._device_counter += 1

    def turn_on(self):
        self.__is_on = True
        print(f"Device {self._device_id} turned on.")

    def turn_off(self):
        self.__is_on = False
        print(f"Device {self._device_id} turned off.")

    @property
    def is_on(self):
        return self.__is_on

    @abstractmethod
    def get_status_report(self):
        pass

    @abstractmethod
    def perform_action(self, action_type, value=None):
        pass

    @classmethod
    def get_device_count(cls):
        return cls._device_counter

    @staticmethod
    def get_system_time():
        return datetime.now().strftime("%Y-%m-%d %H:%M:%S")

```

```python
class SmartLight(SmartDevice):
    def __init__(self, device_id):
        super().__init__(device_id)
        self.__brightness = 0

    @property
    def brightness(self):
        return self.__brightness

    @brightness.setter
    def brightness(self, value):
        if not self.is_on:
            print("Light is not turned on")
            return
        if 0 <= value <= 100:
            self.__brightness = value
            print(f"Brightness set to {value}%.")
        else:
            print("brightness should be between 0 and 100")

    def get_status_report(self):
        return f"SmartLight {self._device_id} - On: {self.is_on}, Brightness: {self.__brightness}%"

    def perform_action(self, action_type, value=None):
        if action_type == "set_brightness":
            self.brightness = value
        else:
            print("Action invalid")

```

```python
class SmartThermostat(SmartDevice):
    def __init__(self, device_id):
        super().__init__(device_id)
        self.__temperature = 20.0

    @property
    def temperature(self):
        return self.__temperature

    @temperature.setter
    def temperature(self, value):
        if not self.is_on:
            print("Thermostat is not turned on")
            return
        if 18.0 <= value <= 30.0:
            self.__temperature = value
            print(f"Temperature set to {value}C.")
        else:
            print("Oops Temperature should be between 18 and 30")

    def get_status_report(self):
        return f"SmartThermostat {self._device_id} - On: {self.is_on}, Temperature: {self.__temperature}C"

    def perform_action(self, action_type, value=None):
        if action_type == "set_temperature":
            self.temperature = value
        else:
            print("Invalid action")

```

```python
class Programmable(ABC):
    def __init__(self):
      print("calling from programmable init")
    def schedule_task(self):
        print("Alarm system task scheduled.")

class SecuritySensor(SmartDevice):
    def __init__(self, device_id):
        super().__init__(device_id)
        self.__is_armed = False

    def arm_sensor(self):
        self.__is_armed = True
        print(f"Security Sensor {self._device_id} armed.")

    def get_status_report(self):
        return f"SecuritySensor {self._device_id} - On: {self.is_on}, Armed: {self.__is_armed}"

    def perform_action(self, action_type, value=None):
        if action_type == "arm":
            self.arm_sensor()
        else:
            print("Action invalid")
```

```python

class SmartAlarmSystem(SmartDevice, Programmable):
    def __init__(self, device_id):
        super().__init__(device_id)
        Programmable()

        self.__is_armed = False

    def get_status_report(self):
        return f"SmartAlarmSystem {self._device_id} - On: {self.is_on}, Armed: {self.__is_armed}"

    def perform_action(self, action_type, value=None):
        if action_type == "arm":
            self.__is_armed = True
            print("Alarm armed.")
        elif action_type == "disarm":
            self.__is_armed = False
            print("Alarm disarmed.")
        else:
            print("Action invalid")

    def schedule_task(self):
        print("Alarm system task scheduled.")

```

```python
class HomeManager:
    def __init__(self):
        self._devices = []

    def add_device(self, device):
        if isinstance(device, SmartDevice):
            self._devices.append(device)
            print(f"Device {device._device_id} added.")
        else:
            print("Oops! Only SmartDevice instances can be added.")

    def control_device(self, device_id, action_type, value=None):
        for device in self._devices:
            if device._device_id == device_id:
                device.perform_action(action_type, value)
                return
        print("Sorry Device not found.")

    def get_all_device_statuses(self):
        for device in self._devices:
            print(device.get_status_report())

```

```python
manager = HomeManager()

print(SmartDevice.get_device_count())


light = SmartLight("1")
thermostat = SmartThermostat("2")
alarm = SmartAlarmSystem("3")

print(light.get_system_time())

manager.add_device(light)
manager.add_device(thermostat)
manager.add_device(alarm)

print(SmartDevice.get_device_count())


light.turn_on()
thermostat.turn_on()
alarm.turn_on()

manager.control_device("1", "set_brightness", 75)
manager.control_device("2", "set_temperature", 25)
manager.control_device("3", "arm")

manager.control_device("1", "set_brightness", 150)

manager.get_all_device_statuses()

light.turn_off()

manager.get_all_device_statuses()

print("\nSmartAlarmSystem MRO:")
print(SmartAlarmSystem.__mro__)
```
