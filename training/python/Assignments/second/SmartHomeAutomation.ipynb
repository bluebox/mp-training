```python
from abc import ABC,abstractmethod
from datetime import datetime
class SmartDevice(ABC):
    count=0
    def __init__(self,device_id):
        self._device_id=device_id
        self.__is_on=False
        SmartDevice.count+=1  
    def turn_on(self):
        self.__is_on=True
        print(f"Device {self._device_id} turned on")    
    def turn_off(self):
        self.__is_on=False
        print(f"Device {self._device_id} turned off")
    @abstractmethod
    def get_status_report():
        pass
    @abstractmethod
    def perform_action(action_type,value=None):
        pass 
    @property
    def is_on(self):
        return self.__is_on
    @classmethod
    def get_device_count(cls):
        return cls.count
    @staticmethod
    def get_system_time():
        return datetime.now()
```

```python
class SmartLight(SmartDevice):
    def __init__(self, device_id):
        super().__init__(device_id)
        self.__brightness=0
    def get_status_report(self):
        return f"Light id: {self._device_id} is_on:{self.is_on} brightness: {self.__brightness}"
    def perform_action(self,action_type,value=None):
        if action_type=="set_brightness":
            if self.is_on and 0<value<100:
                self.__brightness=value
                print("success")
            else:
                print("error")
        else:
            print("invalid action type")
    @property
    def brightness(self):
        return self.__brightness
    @brightness.setter
    def brightness(self,value):
        self.__brightness=value
```

```python
class SmartThermostat(SmartDevice):
    def __init__(self, device_id):
        super().__init__(device_id)
        self.__temperature=20.0   
    def get_status_report(self):
        return f"Thermostat id: {self._device_id} is_on:{self.is_on} temperature: {self.__temperature}"
    def perform_action(self,action_type,value=None):
        if action_type=="set_temperature":
            if self.is_on and 18.0<=value<=30.0:
                self.__temperature=value
                print("success")
            else:
                print("error")
        else:
            print("invalid action type")           
    @property
    def temperature(self):
        return self.__temperature
    @temperature.setter
    def temperature(self,temp):
        self.__temperature=temp
```

```python
class Programmable:
    def scheduled_task(self):
        pass
```

```python
class SecuritySensor(SmartDevice):
    def __init__(self,device_id):
        super().__init__(device_id)
        self.__is_armed=False
    def arm_sensor(self):
        self.__is_armed=True
        print("sensor armed")
```

```python
class SmartAlarmSystem(SmartDevice,Programmable):
    def __init__(self,device_id):
        super().__init__(device_id)
        self.__is_armed=False
    def get_status_report(self):
        return f"alarm id: {self._device_id} is_on:{self.is_on} armed: {self.__is_armed}"
    def perform_action(self,action_type,value=None):
        if action_type=="arm":
            self.__is_armed=True
            print("Armed")
        elif action_type=="disarm":
            self.__is_armed=False
            print("disarmed")
        else:
            print("invalid action type")
    def scheduled_task(self):
        print("alarm system task scheduled")
```

```python
class HomeManager:
    def __init__(self):
        self._devices=[]
    def add_device(self,device):
        self._devices.append(device)
    def control_device(self,device_id,action_type,value=None):
        for device in self._devices:
            if device._device_id==device_id:
                device.perform_action(action_type,value)
                return
        print("no device id")
    def get_all_device_statuses(self):
        for device in self._devices:
            print(device.get_status_report())
```

```python
hm=HomeManager()
```

```python
light=SmartLight("SD001")
thermostat=SmartThermostat("SD002")
alarm=SmartAlarmSystem("SD003")
```

```python
hm.add_device(light)
hm.add_device(thermostat)
hm.add_device(alarm)
```

```python
light.turn_on()
thermostat.turn_on()
alarm.turn_on()
```

```python
hm.control_device("SD001","set_brightness",75)
hm.control_device("SD002","set_temperature",25)
hm.control_device("SD003","arm")
hm.control_device("SD001","set_brightness",150)
```

```python
hm.get_all_device_statuses()
```
